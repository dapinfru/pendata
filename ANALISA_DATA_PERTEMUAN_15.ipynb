# 3. Apa Itu Lag?

## Pengertian Lag

Lag adalah nilai data pada periode sebelumnya yang digunakan sebagai informasi untuk memprediksi nilai pada periode berikutnya.

Dalam forecasting, data masa lalu dianggap memiliki hubungan dengan data masa depan sehingga digunakan sebagai fitur atau variabel input.

### Contoh

| Hari | Demand |
|--------|--------|
| Senin | 100 |
| Selasa | 120 |
| Rabu | 130 |
| Kamis | ? |

Jika ingin memprediksi Demand pada hari Kamis maka:

- lag_1 = 130 (Rabu)
- lag_2 = 120 (Selasa)
- lag_3 = 100 (Senin)

Artinya:

- lag_1 = nilai 1 periode sebelumnya.
- lag_2 = nilai 2 periode sebelumnya.
- lag_3 = nilai 3 periode sebelumnya.

Pada kasus Skforecast digunakan 7 lag sehingga model melihat data Demand selama 7 hari terakhir untuk memprediksi hari berikutnya.
