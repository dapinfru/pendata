## Decision Tree

## 1. Pengertian Decision Tree

Decision Tree atau pohon keputusan adalah salah satu algoritma yang digunakan untuk klasifikasi. Algoritma ini bekerja dengan cara membagi data berdasarkan atribut tertentu sampai menghasilkan keputusan akhir. Bentuk dari model ini menyerupai pohon yang terdiri dari akar, cabang, dan daun.

Pada Decision Tree, setiap node berfungsi untuk melakukan pengujian terhadap suatu atribut. Setiap cabang menunjukkan hasil dari pengujian atribut tersebut, sedangkan leaf node atau daun menunjukkan hasil keputusan akhir berupa kelas tertentu. Algoritma ini mudah dipahami karena proses pengambilan keputusannya dapat ditampilkan dalam bentuk visual.

Dalam proyek ini, algoritma Decision Tree digunakan untuk melakukan klasifikasi pada dataset **Play Tennis**. Target yang digunakan adalah kolom `Play Tennis`, yaitu kolom yang menunjukkan apakah seseorang dapat bermain tenis atau tidak berdasarkan kondisi cuaca tertentu.

## 2. Konsep Dasar Decision Tree

Decision Tree membentuk keputusan dengan cara membagi data menjadi beberapa kelompok berdasarkan atribut yang paling berpengaruh terhadap kelas target. Atribut terbaik dipilih menggunakan ukuran kualitas tertentu, seperti **Gini Index** atau **Gain Ratio**.

Pada proyek ini, ukuran yang digunakan adalah **Gain Ratio**. Gain Ratio digunakan untuk menentukan atribut yang paling tepat sebagai pemisah data atau sebagai root pada pohon keputusan.

### Rumus Gain Ratio

Gain Ratio dihitung dengan membandingkan nilai **Information Gain** dengan **Split Information**. Rumus Gain Ratio adalah sebagai berikut:

$$
\text{Gain Ratio}(S, A) = \frac{\text{Gain}(S, A)}{\text{SplitInfo}(S, A)}
$$

Sebelum menghitung Gain Ratio, terlebih dahulu perlu dihitung nilai **Entropy** dan **Information Gain**.

Rumus Entropy:

$$
\text{Entropy}(S) = - \sum p_i \log_2(p_i)
$$

Rumus Information Gain:

$$
\text{Gain}(S, A) = \text{Entropy}(S) - \sum \frac{|S_v|}{|S|} \times \text{Entropy}(S_v)
$$

Rumus Split Information:

$$
\text{SplitInfo}(S, A) = - \sum \frac{|S_v|}{|S|} \log_2\left(\frac{|S_v|}{|S|}\right)
$$

Keterangan:

- **S**: seluruh data yang digunakan.
- **A**: atribut yang diuji untuk membagi data.
- **Sv**: subset data berdasarkan nilai tertentu pada atribut A.
- **pi**: proporsi data pada kelas ke-i.
- **|Sv| / |S|**: perbandingan jumlah data pada subset terhadap total data.

Atribut yang memiliki nilai **Gain Ratio** paling tinggi akan dipilih sebagai atribut terbaik untuk membentuk percabangan pada Decision Tree.

## 3. Struktur Decision Tree

Struktur utama pada Decision Tree terdiri dari:

1. **Root Node**  
   Root node merupakan node paling awal atau akar dari pohon keputusan. Node ini berisi atribut pertama yang digunakan untuk membagi data.

2. **Internal Node**  
   Internal node adalah node yang berada di bagian tengah pohon dan digunakan untuk menguji atribut lain.

3. **Branch**  
   Branch atau cabang menunjukkan hasil dari pengujian suatu atribut.

4. **Leaf Node**  
   Leaf node merupakan node akhir yang menunjukkan hasil keputusan atau kelas prediksi.

## 4. Kelebihan Decision Tree

Beberapa kelebihan dari algoritma Decision Tree adalah:

- Mudah dipahami dan divisualisasikan.
- Dapat digunakan pada data numerik maupun kategorikal.
- Proses pengambilan keputusan dapat dilihat dengan jelas.
- Tidak memerlukan proses normalisasi data.
- Cocok digunakan untuk menjelaskan pola keputusan dari suatu dataset.

## 7. Membaca Data dan Membuat Workflow Decision Tree di KNIME

Pada bagian ini, proses pembuatan workflow dilakukan menggunakan beberapa node yang saling terhubung di KNIME. Workflow tersebut digunakan untuk membaca data, membagi data menjadi data latih dan data uji, membangun model Decision Tree, melakukan prediksi, serta menilai hasil klasifikasi.

### Alur Workflow

Berdasarkan workflow pada gambar, tahapan yang dilakukan adalah sebagai berikut:

1. **Table Reader**  
   Node ini digunakan untuk membaca dataset yang akan digunakan dalam proses klasifikasi. Data yang dibaca berisi atribut-atribut kondisi cuaca dan kolom target `Play Tennis`.

2. **Table Partitioner**  
   Setelah data berhasil dibaca, data kemudian dibagi menjadi dua bagian, yaitu data training dan data testing. Data training digunakan untuk membangun model, sedangkan data testing digunakan untuk menguji hasil prediksi model.

3. **Color Manager**  
   Node ini digunakan untuk memberikan warna berdasarkan kelas target. Pewarnaan ini membantu dalam proses visualisasi hasil Decision Tree agar setiap kelas dapat dibedakan dengan lebih jelas.

4. **Decision Tree Learner**  
   Node ini berfungsi untuk membangun model Decision Tree berdasarkan data training. Pada tahap ini, algoritma akan mempelajari pola dari data dan membentuk aturan keputusan dalam bentuk pohon.

5. **Color Appender**  
   Node ini digunakan untuk menambahkan informasi warna pada data testing. Dengan adanya warna, hasil prediksi dapat divisualisasikan dengan lebih mudah.

6. **Decision Tree Predictor**  
   Node ini digunakan untuk melakukan prediksi terhadap data testing menggunakan model Decision Tree yang telah dibuat sebelumnya. Hasil dari node ini berupa kelas prediksi, yaitu apakah data termasuk ke dalam kelas `Yes` atau `No`.

7. **Scorer**  
   Node Scorer digunakan untuk mengevaluasi hasil prediksi. Node ini membandingkan kelas asli dengan kelas hasil prediksi, sehingga dapat diketahui seberapa baik model Decision Tree dalam melakukan klasifikasi.

### Penjelasan Singkat

Workflow ini dimulai dari proses membaca dataset menggunakan **Table Reader**, kemudian data dibagi menggunakan **Table Partitioner** menjadi data latih dan data uji. Data latih digunakan oleh **Decision Tree Learner** untuk membangun model pohon keputusan. Setelah model terbentuk, data uji diprediksi menggunakan **Decision Tree Predictor**. Hasil prediksi kemudian dievaluasi menggunakan **Scorer** untuk mengetahui performa model.

Dengan workflow ini, proses klasifikasi dataset **Play Tennis** dapat dilakukan secara terstruktur mulai dari pembacaan data hingga evaluasi hasil prediksi.

## 8. Membagi Data Menggunakan Table Partitioner

**Node:** Table Partitioner

**Fungsi:**  
Node **Table Partitioner** digunakan untuk membagi dataset menjadi dua bagian, yaitu data training dan data testing. Pembagian data ini dilakukan sebelum proses pembentukan model pada node **Decision Tree Learner**.

**Konfigurasi:**

- **First partition type:** Relative (%)
- **Relative size:** 90
- **Sampling strategy:** Stratified
- **Group column:** Play Tennis
- **Fixed random seed:** 0
- **If input table is empty:** Fail

Berdasarkan pengaturan tersebut, data dibagi menggunakan tipe pembagian **Relative (%)** dengan ukuran partisi pertama sebesar **90%**. Artinya, sebanyak 90% data digunakan sebagai data training, sedangkan sisa data digunakan sebagai data testing.

Strategi sampling yang digunakan adalah **Stratified**. Strategi ini membagi data dengan tetap mempertahankan perbandingan jumlah kelas pada kolom target. Pada workflow ini, kolom yang digunakan sebagai acuan stratifikasi adalah `Play Tennis`.

Penggunaan **Fixed random seed** dengan nilai **0** bertujuan agar hasil pembagian data tetap sama setiap kali workflow dijalankan ulang. Dengan demikian, proses eksperimen menjadi lebih konsisten dan hasilnya dapat dibandingkan dengan lebih mudah.

Jika tabel input kosong, maka pengaturan **Fail** akan membuat node berhenti dan menampilkan kesalahan. Hal ini berguna untuk memastikan bahwa proses selanjutnya hanya berjalan ketika data tersedia.

## 9. Membuat Model Menggunakan Decision Tree Learner

**Node:** Decision Tree Learner

**Fungsi:**  
Node **Decision Tree Learner** digunakan untuk membentuk model klasifikasi menggunakan algoritma Decision Tree. Data yang digunakan berasal dari hasil pembagian pada node **Table Partitioner**, khususnya data training. Node ini akan mempelajari pola dari setiap atribut pada dataset untuk menghasilkan aturan keputusan yang dapat memprediksi kelas pada kolom target.

**Konfigurasi:**

- **Class column:** Play Tennis
- **Quality measure:** Gain ratio
- **Pruning method:** No pruning
- **Reduced error pruning:** Aktif
- **Minimum number of records per node:** 2
- **Number of records to store for view:** 10000
- **Average split point:** Aktif
- **Number threads:** 12
- **Skip nominal columns without domain information:** Aktif

Pada konfigurasi ini, kolom `Play Tennis` dipilih sebagai **class column** karena kolom tersebut merupakan target yang akan diprediksi oleh model. Dengan kata lain, model akan belajar dari atribut-atribut lain untuk menentukan apakah hasil akhirnya termasuk kelas `Yes` atau `No`.

Ukuran kualitas yang digunakan adalah **Gain ratio**. Metode ini berfungsi untuk menentukan atribut terbaik yang akan dijadikan percabangan dalam pohon keputusan. Atribut dengan nilai gain ratio paling baik akan dipilih untuk membagi data pada setiap node.

Pada bagian **Pruning method**, digunakan pilihan **No pruning**, sehingga pohon keputusan tidak dipangkas secara khusus pada tahap tersebut. Namun, opsi **Reduced error pruning** tetap diaktifkan untuk membantu mengurangi kemungkinan model terlalu menyesuaikan diri terhadap data training.

Pengaturan **Minimum number of records per node** bernilai **2**, yang berarti setiap node minimal harus memiliki dua data agar dapat diproses lebih lanjut. Pengaturan ini membantu agar struktur pohon yang terbentuk tidak terlalu kompleks.

Nilai **Number of records to store for view** diatur sebesar **10000**, sehingga KNIME dapat menyimpan sejumlah data tersebut untuk kebutuhan tampilan hasil model. Opsi **Average split point** diaktifkan untuk membantu menentukan titik pemisahan pada atribut numerik.

Selain itu, **Number threads** diatur menjadi **12** agar proses pembentukan model dapat berjalan lebih cepat dengan memanfaatkan beberapa thread. Opsi **Skip nominal columns without domain information** juga diaktifkan agar kolom nominal yang tidak memiliki informasi domain dapat dilewati dalam proses pembelajaran model.

## 13. Hasil dan Pembahasan

Berdasarkan workflow yang telah dibuat di KNIME, proses klasifikasi menggunakan algoritma **Decision Tree** berhasil dilakukan dengan baik. Dari hasil tersebut, dapat diketahui bahwa model Decision Tree mampu membentuk aturan keputusan yang sederhana dan mudah dipahami. Visualisasi dalam bentuk pohon memudahkan pengguna untuk melihat alur pengambilan keputusan berdasarkan atribut-atribut yang terdapat pada dataset.
