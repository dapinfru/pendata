## Konsep dan Implementasi Data Science

Data Science adalah bidang ilmu yang mempelajari proses pengumpulan, pengolahan, analisis, dan interpretasi data untuk menghasilkan informasi yang berguna dalam pengambilan keputusan. # Metodologi CRISP-DM

CRISP-DM (*Cross Industry Standard Process for Data Mining*) adalah metodologi standar yang digunakan dalam pengerjaan proyek Data Science dan Data Mining. Metode ini membantu proses analisis data menjadi lebih terstruktur, sistematis, dan mudah dipahami.

CRISP-DM terdiri dari beberapa tahapan utama yang saling berhubungan. Setiap tahap memiliki tujuan tertentu agar proyek Data Science dapat berjalan dengan baik dan menghasilkan solusi yang sesuai dengan kebutuhan bisnis.

Tahapan dalam CRISP-DM:
1. Business Understanding
2. Data Understanding
3. Data Preprocessing
4. Modelling
5. Evaluation
6. Deployment

---

# 1. Business Understanding

Tahap pertama dalam CRISP-DM adalah memahami permasalahan bisnis yang ingin diselesaikan. Pada tahap ini, fokus utama bukan pada data, tetapi memahami tujuan bisnis dan kebutuhan perusahaan.

Tujuan tahap ini:
- Menentukan masalah yang ingin diselesaikan
- Menentukan tujuan proyek
- Menentukan kebutuhan analisis
- Membuat rencana pengerjaan proyek

Kegiatan yang dilakukan:
- Diskusi dengan pihak bisnis atau perusahaan
- Menentukan target yang ingin dicapai
- Menentukan indikator keberhasilan proyek
- Menentukan sumber daya dan jadwal proyek

Contoh kasus:
Sebuah bank ingin memprediksi apakah nasabah akan gagal membayar kredit (*credit risk modelling*). Dengan analisis data, bank dapat menentukan:
- Nasabah yang layak menerima pinjaman
- Besar pinjaman yang diberikan
- Risiko keterlambatan pembayaran

Tahap ini sangat penting karena kesalahan memahami masalah bisnis dapat menyebabkan model yang dibuat tidak sesuai kebutuhan perusahaan.

# Jenis Atribut Data

Dalam Data Science, atribut atau fitur merupakan karakteristik yang dimiliki oleh data. Setiap atribut memiliki tipe yang berbeda sehingga penanganannya juga berbeda.

---

# 1. Categorical Attribute

Atribut kategorikal adalah atribut yang nilainya berupa kategori atau label tertentu.

Ciri-ciri:
- Berupa label atau nama
- Tidak memiliki urutan
- Tidak dapat dihitung secara matematis

Contoh:
- Jenis kelamin
- Kota
- Warna
- Status pernikahan

Contoh data:

| Nama | Jenis Kelamin |
|------|----------------|
| Andi | Laki-laki |
| Sinta | Perempuan |

Pada data kategorikal, nilai hanya menunjukkan kategori dan tidak memiliki arti numerik.

---

# 2. Ordinal Attribute

Atribut ordinal adalah atribut kategorikal yang memiliki tingkatan atau urutan tertentu.

Ciri-ciri:
- Memiliki urutan
- Tidak memiliki jarak yang pasti antar kategori

Contoh:
- Rendah
- Sedang
- Tinggi

Contoh lain:
- Tingkat pendidikan:
  - SMA
  - S1
  - S2
  - S3

Walaupun memiliki urutan, selisih antar kategori tidak dapat diukur secara pasti.

---

# 3. Numeric Attribute

Atribut numerik adalah atribut yang nilainya berupa angka dan dapat dilakukan operasi matematika.

Atribut numerik dibagi menjadi dua jenis:

## a. Numeric Diskrit

Data numerik diskrit berupa bilangan bulat dan biasanya digunakan untuk menghitung jumlah.

Contoh:
- Jumlah anak
- Jumlah kendaraan
- Jumlah transaksi

Contoh data:

| Nama | Jumlah Anak |
|------|--------------|
| Andi | 2 |
| Budi | 3 |

---

## b. Numeric Kontinu

Data numerik kontinu berupa angka pecahan atau bilangan real.

Contoh:
- Tinggi badan
- Berat badan
- Jarak
- Suhu

Contoh data:

| Nama | Tinggi Badan |
|------|----------------|
| Andi | 170.5 |
| Sinta | 158.2 |

Data kontinu biasanya memiliki nilai dalam rentang tertentu.

# Data Quality

Kualitas data sangat penting dalam proses Data Science karena kualitas model sangat bergantung pada kualitas data yang digunakan.

Konsep penting:
> Garbage In, Garbage Out

Artinya:
Jika data yang digunakan buruk, maka hasil analisis dan model yang dihasilkan juga akan buruk.

Dimensi kualitas data:

## 1. Akurasi
Data harus sesuai dengan kondisi sebenarnya.

Contoh:
Jika umur seseorang 20 tahun tetapi tercatat 200 tahun, maka data tersebut tidak akurat.

---

## 2. Kelengkapan
Data tidak boleh banyak yang kosong (*missing value*).

Contoh:
Kolom penghasilan banyak yang kosong sehingga analisis menjadi sulit.

---

## 3. Konsistensi
Format data harus seragam.

Contoh:
- "L"
- "Laki-laki"
- "Pria"

Ketiga nilai tersebut memiliki arti sama tetapi formatnya berbeda.

---

## 4. Ketepatan Waktu
Data harus relevan dengan kondisi terbaru.

Contoh:
Data perilaku pelanggan tahun 2018 mungkin tidak relevan untuk kondisi saat ini.

---

# Missing Value

Missing value adalah kondisi ketika data memiliki nilai kosong atau tidak terisi.

Penyebab missing value:
- Kesalahan input data
- Sensor rusak
- Data tidak diisi
- Data hilang saat proses penyimpanan

Contoh data:

| Nama | Umur |
|------|------|
| Andi | 20 |
| Budi | NaN |

Nilai `NaN` menunjukkan data kosong atau missing value.

Missing value dapat menyebabkan:
- Analisis menjadi tidak akurat
- Model machine learning gagal bekerja
- Hasil prediksi menjadi buruk

Oleh karena itu missing value harus ditangani sebelum proses modelling.

---

# Cara Mengatasi Missing Value

Terdapat beberapa cara untuk menangani missing value.

---

# 1. Menghapus Data

Data yang memiliki missing value dapat dihapus.

Bisa berupa:
- Menghapus baris
- Menghapus kolom

Kelebihan:
- Mudah dilakukan
- Data menjadi bersih

Kelemahan:
- Mengurangi jumlah data
- Bisa menghilangkan informasi penting

Metode ini cocok jika jumlah missing value sedikit.

---

# 2. Imputasi Missing Value

Imputasi adalah proses mengisi nilai kosong menggunakan nilai tertentu.

Metode imputasi yang umum:

## a. Mean
Mengisi missing value menggunakan rata-rata data.

Cocok untuk:
- Data numerik

---

## b. Median
Mengisi missing value menggunakan nilai tengah data.

Cocok jika:
- Data memiliki outlier

---

## c. Modus
Mengisi missing value menggunakan nilai yang paling sering muncul.

Cocok untuk:
- Data kategorikal

---

# Contoh Imputasi Mean Menggunakan Python

```python
import pandas as pd
import numpy as np

# Membuat dataframe
df = pd.DataFrame({
    "Nilai": [10, 20, np.nan, 40]
})

print("Data Sebelum Imputasi")
print(df)

# Mengisi missing value dengan mean
df["Nilai"] = df["Nilai"].fillna(df["Nilai"].mean())

print("\nData Setelah Imputasi")
print(df)
